```python
class Encapsulated:
    def __init__(self, value):
        self.__value = None
        self.value = value

    # Your code here (@property & @value.setter)
    @property
    def value(self):
        print("__property__")
        return self.__value

    @value.setter
    def value(self, val):
        print("__setter__")
        if (not isinstance(val, int)):
            raise ValueError("Значение должено быть int")
        self.__value = val


def test1():
    obj = Encapsulated(10)
    print("Test 1:", obj.value)
    print("Test 1:", obj._Encapsulated__value)


def test2():
    try:
        obj = Encapsulated("hello")
        print("Test 2:", obj.value)
        print("Test 2:", obj._Encapsulated__value)
    except ValueError as ve:
        print("Test 2: OK")


def test3():
    try:
        obj = Encapsulated(10)
        print("Test 3:", obj.value)
        print("Test 3:", obj._Encapsulated__value)
    except ValueError as ve:
        print("Test 3: OK")


def test4():
    obj = Encapsulated(10)
    print("Test 4:", obj.value)
    print("Test 4:", obj._Encapsulated__value)

    obj.value = 5
    print("Test 4:", obj.value)
    print("Test 4:", obj._Encapsulated__value)


test1()
print()

test2()
print()

test3()
print()

test4()
print()
```

## Задание 7

Напишите класс с поддержкой множественного наследования, демонстрирующий работу C3-линеаризации MRO на сложном примере с 3+ уровнями наследования и пересечениями.

```python
class A:
   def method(self):
      print("A")

class B(A):
   def method(self):
      print("B")
      super().method()

class C(A):
   def method(self):
      print("C")
      super().method()

class D(B, C):
   def method(self):
      print("D")
      super().method()

d = D()
d.method()
print(D.__mro__)
```

```python
class LifeCycle:
   def __new__(cls):
        print("__new__ ->", end=" ")
        return super().__new__(cls)

   def __init__(self):
      print("__init__ ->", end=" ")

   def __call__(self):
      print("__call__ ->", end=" ")

   def __del__(self):
      print("__del__")

   def __repr__(self):
      return "__repr__ ->"

   def __str__(self):
      return "__str__ ->"

print("Последовательность следующая:")
obj = LifeCycle()
print(obj, end=" ")
obj()
del obj
```

## Задание 9

Создайте контекстный менеджер с помощью специальных методов enter и exit, используйте его вместе с классом, в котором присутствуют методы с разграничением прав доступа.

https://habr.com/ru/articles/739326/

```python
class SecretData:
    def __init__(self):
        self.__secret = "*приватный ключ*"
    
    @property
    def secret(self):
        raise AttributeError("Доступ к секрету запрещен!")


class SecretContext:
    def __init__(self, data: SecretData):
        self._data = data

    def __enter__(self):
        print("Вход в контекст")
        return self._data._SecretData__secret

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Выход из контекста")


data = SecretData()
ctx = SecretContext(data)

try:
    print(data.secret)
except AttributeError as e:
    print(e)

with ctx as secret:
    print(f"Секрет внутри контекста: {secret}")

try:
    print(data.secret)
except AttributeError as e:
    print(e)
```

Создайте класс, объекты которого могут быть отслежены с помощью слабых ссылок (weakref). Реализуйте систему, которая хранит слабые ссылки на все созданные объекты класса и автоматически удаляет их из списка при уничтожении объектов. Продемонстрируйте это поведение, выводя текущее количество живых объектов.

```python
import weakref

class CountWeak:
    _instances = weakref.WeakSet()

    def __init__(self, name):
        self.name = name
        self._instances.add(self)

    @classmethod
    def live_count(cls):
        return len(cls._instances)

a = CountWeak("A")
b = CountWeak("B")
print(CountWeak.live_count())  # 2

del a
print(CountWeak.live_count())  # 1
```
